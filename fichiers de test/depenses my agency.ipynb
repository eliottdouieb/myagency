````python
import os
import json
from openai import OpenAI

# ============================================================
# 1. Configuration du client OpenAI
# ============================================================
# Assure-toi d'avoir défini la variable d'environnement :
# export OPENAI_API_KEY="sk-XXXX"
client = OpenAI(api_key="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")




# ============================================================
# 3. Construction du prompt
# ============================================================
def build_prompt(revolut_labels, backoffice_labels):
    return f"""
Tu es un assistant spécialisé en rapprochement comptable.

On te donne :
1) La liste des libellés Revolut (revolut_labels).
2) La liste des libellés Back Office (backoffice_labels).

Objectif :
Pour chaque libellé Revolut, tu dois trouver le libellé Back Office le plus probable.
Si tu n'es pas sûr ou qu'il n'y a pas de correspondance raisonnable, retourne "match non trouve".

Contraintes :
- Retourne UNIQUEMENT un JSON valide, sans texte autour.
- Le JSON doit être un objet Python de la forme :
{{
  "Label Revolut 1": "Label BO ou match non trouve",
  "Label Revolut 2": "Label BO ou match non trouve",
  ...
}}
- Ne mets pas de commentaires, pas de texte avant ou après, seulement le JSON.

Voici les libellés Revolut (revolut_labels) :
{json.dumps(revolut_labels, ensure_ascii=False, indent=2)}

Voici les libellés Back Office (backoffice_labels) :
{json.dumps(backoffice_labels, ensure_ascii=False, indent=2)}

Génère maintenant le JSON de mapping.
"""


# ============================================================
# 4. Appel à l'API ChatGPT
# ============================================================
def generate_match_libelle(revolut_labels, backoffice_labels):
    prompt = build_prompt(revolut_labels, backoffice_labels)

    response = client.chat.completions.create(
        model="gpt-4.1-mini",  # ou un autre modèle de ton choix
        messages=[
            {"role": "system", "content": "Tu es un assistant expert en rapprochement comptable."},
            {"role": "user", "content": prompt},
        ],
        temperature=0.1,
    )

    raw_content = response.choices[0].message.content.strip()

    # ----- DEBUG éventuel -----
    # print("=== Réponse brute du modèle ===")
    # print(raw_content)
    # ---------------------------

    # 1) Si la réponse est encadrée par ```...```, on enlève les fences Markdown
    if raw_content.startswith("```"):
        # On coupe sur les ``` pour enlever le cadre de code
        parts = raw_content.split("```")
        # Exemple : ['', 'json\n{...}\n', '']
        if len(parts) >= 2:
            raw_content = parts[1]  # on prend le bloc central
        # On enlève éventuellement le mot "json" en tête de ligne
        raw_content = raw_content.lstrip()
        if raw_content.lower().startswith("json"):
            raw_content = raw_content.split("\n", 1)[1].lstrip()

    # 2) On tente de parser le JSON
    try:
        match_libelle = json.loads(raw_content)
    except json.JSONDecodeError:
        # Si ça plante encore, on affiche pour comprendre
        print("⚠️ Réponse non parseable en JSON, contenu brut :")
        print(raw_content)
        raise

    return match_libelle



````

````python
import pandas as pd

# =========================
# 1. Préparation BACKOFFICE
# =========================
df_bo_clean = df_bo.copy()

# Date au bon format
df_bo_clean["Date"] = pd.to_datetime(df_bo_clean["Date"], errors="coerce")

# Montant signé : Débit - Crédit
df_bo_clean["Montant"] = df_bo_clean["Débit(€)"] - df_bo_clean["Crédit (€)"]

# On ne garde que les montants positifs
df_bo_clean = df_bo_clean[df_bo_clean["Montant"] > 0]

# On garde un identifiant stable des lignes BO
df_bo_clean = df_bo_clean.reset_index().rename(columns={"index": "idx_bo"})


# ======================
# 2. Préparation REVOLUT
# ======================
df_rev_clean = df.copy()

# On ne garde que les paiements CB
df_rev_clean = df_rev_clean[df_rev_clean["Type"] == "CARD_PAYMENT"]

# Date de référence = Date started (UTC)
df_rev_clean["Date"] = pd.to_datetime(
    df_rev_clean["Date started (UTC)"],
    errors="coerce"
)

# Montant déjà signé
df_rev_clean["Montant"] = pd.to_numeric(
    df_rev_clean["Amount"]*(-1),
    errors="coerce"
)

# Colonnes utiles
cols_rev_keep = [
    "Date",
    "Montant",
    "Description",
    "ID",
    "Type",
    "State",
    "Card number",
    "Card label",
    "Payer",
    "Exchange rate",
    "Orig currency",
    "Orig amount",
    "email"
]
df_rev_clean = df_rev_clean[cols_rev_keep]

# Libellé BO attendu (mapping via ton dict match_libelle)
df_rev_clean["Libelle_match"] = df_rev_clean["Description"].map(match_libelle)

# Identifiant stable des lignes Revolut
df_rev_clean = df_rev_clean.reset_index().rename(columns={"index": "idx_rev"})


# =========================
# 3. Helpers pour éviter les doublons
# =========================
used_rev = set()  # idx_rev déjà matchés
used_bo = set()   # idx_bo déjà matchés

def filtre_nouveaux(df):
    """Ne garde que les couples idx_rev / idx_bo non encore utilisés."""
    return df[
        ~df["idx_rev"].isin(used_rev)
        & ~df["idx_bo"].isin(used_bo)
    ]

def maj_sets(df):
    """Met à jour les sets de lignes déjà utilisées."""
    used_rev.update(df["idx_rev"].dropna().unique())
    used_bo.update(df["idx_bo"].dropna().unique())


# =========================================
# 4. Match exact : Date + Montant + Libellé
# =========================================
matches_ok = df_rev_clean.merge(
    df_bo_clean,
    left_on=["Date", "Montant", "Libelle_match"],
    right_on=["Date", "Montant", "Libelle"],
    how="inner",            # on ne veut que les deux côtés
    suffixes=("_rev", "_bo")
)

matches_ok = matches_ok.drop_duplicates(subset=["idx_rev", "idx_bo"])
maj_sets(matches_ok)

print("🔹 Lignes matchées (Montant, Libellé, Date) :")
display(
    matches_ok[
        [
            "Date",
            "Libelle",
            "Débit(€)",
            "Crédit (€)",
            "Montant",
            "Description",
            "ID",
            "Type",
            "State",
            "Payer",
            "Exchange rate",
            "Orig currency",
            "Orig amount"
        ]
    ]
)


# =====================================
# 5. Match sans libellé (Date + Montant)
# =====================================
matches_sans_libelle = df_rev_clean.merge(
    df_bo_clean,
    left_on=["Date", "Montant"],
    right_on=["Date", "Montant"],
    how="inner",
    suffixes=("_rev", "_bo")
)

matches_sans_libelle = matches_sans_libelle.drop_duplicates(subset=["idx_rev", "idx_bo"])
matches_sans_libelle = filtre_nouveaux(matches_sans_libelle)
maj_sets(matches_sans_libelle)

print("🔹 Lignes matchées sans libellé (Montant, Date) :")
display(
    matches_sans_libelle[
        [
            "Date",
            "Libelle",
            "Débit(€)",
            "Crédit (€)",
            "Montant",
            "Description",
            "ID",
            "Type",
            "State",
            "Payer",
            "Exchange rate",
            "Orig currency",
            "Orig amount"
        ]
    ]
)


# ====================================
# 6. Match sans date (Montant + Libellé)
# ====================================
matches_sans_date = df_rev_clean.merge(
    df_bo_clean,
    left_on=["Montant", "Libelle_match"],
    right_on=["Montant", "Libelle"],
    how="inner",
    suffixes=("_rev", "_bo")
)

matches_sans_date = matches_sans_date.drop_duplicates(subset=["idx_rev", "idx_bo"])
matches_sans_date = filtre_nouveaux(matches_sans_date)
maj_sets(matches_sans_date)

print("🔹 Lignes matchées sans date (Montant, Libellé) :")
display(
    matches_sans_date[
        [
            "Date_rev",
            "Date_bo",
            "Libelle",
            "Débit(€)",
            "Crédit (€)",
            "Montant",
            "Description",
            "ID",
            "Type",
            "State",
            "Payer",
            "Exchange rate",
            "Orig currency",
            "Orig amount"
        ]
    ]
)


# =====================================
# 7. Match sans montant (Date + Libellé)
# =====================================
matches_sans_montant = df_rev_clean.merge(
    df_bo_clean,
    left_on=["Date", "Libelle_match"],
    right_on=["Date", "Libelle"],
    how="inner",
    suffixes=("_rev", "_bo")
)

matches_sans_montant = matches_sans_montant.drop_duplicates(subset=["idx_rev", "idx_bo"])
matches_sans_montant = filtre_nouveaux(matches_sans_montant)
maj_sets(matches_sans_montant)

print("🔹 Lignes matchées sans montant (Date, Libellé) :")
display(
    matches_sans_montant[
        [
            "Date",
            "Libelle",
            "Débit(€)",
            "Crédit (€)",
            "Montant_rev",
            "Montant_bo",
            "Description",
            "ID",
            "Type",
            "State",
            "Payer",
            "Exchange rate",
            "Orig currency",
            "Orig amount"
        ]
    ]
)


# =======================================================
# 8. Potentiels : même libellé, date à ± 3 jours
# =======================================================
matches_potentiel = df_rev_clean.merge(
    df_bo_clean,
    left_on=["Libelle_match"],
    right_on=["Libelle"],
    how="inner",
    suffixes=("_rev", "_bo")
)

matches_potentiel = matches_potentiel.drop_duplicates(subset=["idx_rev", "idx_bo"])

# Calcul de l'écart de jours entre les deux dates
matches_potentiel["ecart_jours"] = (
    (matches_potentiel["Date_bo"] - matches_potentiel["Date_rev"])
    .dt.days
    .abs()
)

# On ne garde que les écarts <= 3 jours
matches_potentiel = matches_potentiel[matches_potentiel["ecart_jours"] <= 3]

# On exclut ceux déjà utilisés
matches_potentiel = filtre_nouveaux(matches_potentiel)
maj_sets(matches_potentiel)

print("🔹 Lignes matchées POTENTIELLES (Date ±3 jours, Libellé) :")
display(
    matches_potentiel[
        [
            "Date_rev",
            "Date_bo",
            "Libelle",
            "Montant_rev",
            "Montant_bo",
            "Description",
            "ID",
            "Type",
            "State",
            "Payer",
            "Exchange rate",
            "Orig currency",
            "Orig amount"
        ]
    ]
)


# ======================
# 9. Lignes KO (non matchées)
# ======================

# KO Revolut = lignes Revolut jamais utilisées dans aucun match
matches_ko_rev = df_rev_clean[~df_rev_clean["idx_rev"].isin(used_rev)]

# KO BO = lignes Backoffice jamais utilisées
matches_ko_bo = df_bo_clean[~df_bo_clean["idx_bo"].isin(used_bo)]

print("🔹 Lignes matches KO Revolut :")
display(
    matches_ko_rev[
        [
            "Date",
            "Libelle_match",
            "Montant",
            "Description",
            "ID",
            "Type",
            "State",
            "Payer",
            "Exchange rate",
            "Orig currency",
            "Orig amount",
            "email"
        ]
    ]
)

print("🔹 Lignes matches KO Back Office :")
display(
    matches_ko_bo[
        [
            "Date",
            "Libelle",
            "Débit(€)",
            "Crédit (€)",
            "Montant",
            "Compte",
        ]
    ]
)
````

````python
import gspread
import pandas as pd

# =======================================================
# 10. Export des KO Revolut vers Google Sheet (Conciergerie)
# =======================================================

def export_ko_vers_gsheet(df_ko, json_keyfile, sheet_name, worksheet_name="Feuille 1"):
    """
    Prend le dataframe des KO, le formate et l'ajoute au Google Sheet.
    Utilise la méthode native de gspread (plus besoin de oauth2client).
    """
    if df_ko.empty:
        print("✅ Aucune dépense manquante à exporter.")
        return

    print(f"🚀 Export de {len(df_ko)} lignes vers Google Sheet...")

    # 1. Authentification Google (Méthode moderne)
    try:
        # gspread gère l'auth directement maintenant
        gc = gspread.service_account(filename=json_keyfile)
        
        # 2. Ouverture du fichier
        sh = gc.open(sheet_name)
        
        # Sélection de l'onglet
        try:
            ws = sh.worksheet(worksheet_name) 
        except gspread.WorksheetNotFound:
            # Si l'onglet n'existe pas, on prend le premier par défaut
            print(f"⚠️ Onglet '{worksheet_name}' introuvable, utilisation du premier onglet.")
            ws = sh.get_worksheet(0)

    except Exception as e:
        print(f"❌ Erreur de connexion Google Sheet : {e}")
        return

    # 3. Préparation des données pour l'export
    df_export = df_ko.copy()

    # Sélection des colonnes pertinentes
    cols_export = [
        "Date", 
        "Description", 
        "Montant", 
        "ID", 
        "Payer",
        "Exchange rate",
        "Orig currency",
        "Orig amount",
        "email"
    ]
    
    # On ne garde que les colonnes qui existent vraiment
    cols_final = [c for c in cols_export if c in df_export.columns]
    df_export = df_export[cols_final]

    # Nettoyage des données pour l'API Google
    # Dates -> String
    if "Date" in df_export.columns:
        df_export["Date"] = df_export["Date"].dt.strftime("%Y-%m-%d")
    
    # NaN -> Vide
    df_export = df_export.fillna("")

    # Conversion en liste
    data_to_append = df_export.values.tolist()

    # 4. Envoi des données (Append)
    try:
        ws.append_rows(data_to_append)
        print(f"✅ Export terminé ! {len(data_to_append)} lignes ajoutées dans '{sheet_name}'.")
    except Exception as e:
        print(f"❌ Erreur lors de l'écriture dans le Sheet : {e}")

# ==========================
# Exécution
# ==========================

FICHIER_JSON_AUTH = "credentials.json"   # Votre fichier JSON
NOM_DU_GSHEET = "Suivi Dépenses Conciergerie" # Nom du GSheet
NOM_ONGLET = "A traiter"                 # Nom de l'onglet

export_ko_vers_gsheet(
    df_ko=matches_ko_rev, 
    json_keyfile=FICHIER_JSON_AUTH, 
    sheet_name=NOM_DU_GSHEET,
    worksheet_name=NOM_ONGLET
)
````
